#### Leetcode 104. Maximum Depth of Binary Tree

Given the root of a binary tree, return its maximum depth.

A binary tree's maximum depth is the number of nodes along the longest path from the root node down to the farthest leaf node.

 

Example 1:


Input: root = [3,9,20,null,null,15,7]
Output: 3
Example 2:

Input: root = [1,null,2]
Output: 2
 

Constraints:

The number of nodes in the tree is in the range [0, 104].
-100 <= Node.val <= 100

#### Using DFS

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def maxDepth(self, root) -> int:
        if root is None:
            return 0
        left_depth = self.maxDepth(root.left)
        right_depth = self.maxDepth(root.right)
        depth = max(left_depth,right_depth) + 1
        return depth
        
```

### Using BFS

```python
class Solution:
    def maxDepth(self,root) -> int:
        from collections import deque
        if root is None:
            return 0

        queue = deque([root])
        depth = 0

        while queue:
            level_size = len(queue)

            for _ in range(level_size):
                node = queue.popleft()

                if node.left:
                    queue.append(node.left)

                if node.right:
                    queue.append(node.right)

            depth += 1

        return depth
```

#### Leetcode 111. Minimum Depth of Binary Tree

Given a binary tree, find its minimum depth.

The minimum depth is the number of nodes along the shortest path from the root node down to the nearest leaf node.

Note: A leaf is a node with no children.

 

Example 1:


Input: root = [3,9,20,null,null,15,7]
Output: 2
Example 2:

Input: root = [2,null,3,null,4,null,5,null,6]
Output: 5
 

Constraints:

The number of nodes in the tree is in the range [0, 105].
-1000 <= Node.val <= 1000

## Using DFS

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def minDepth(self, root) -> int:
        if root is None:
            return 0
        left = self.minDepth(root.left)
        right = self.minDepth(root.right)

        if left == 0:
            min_depth = right + 1
        elif right == 0:
            min_depth = left + 1
        else:
            min_depth = min(left,right) + 1
        return min_depth
```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def minDepth(self, root) -> int:
        if root is None:
            return 0
        from collections import deque
        queue = deque([root])
        depth = 0

        while queue:
            level_size = len(queue)

            for _ in range(level_size):
                node = queue.popleft()
                if node.left is None and node.right is None:
                    return depth + 1
                if node.left:
                    queue.append(node.left)
                if node.right:
                    queue.append(node.right)


            depth += 1

        return depth
        

```

Input: root = [1,2,2,null,3,null,3]

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def mirror(self,left_root,right_root) -> bool:
        left = True
        right = True
        if left_root is None and right_root is None:
            return True

        if left_root is None and right_root is not None or left_root is not None and right_root is None or left_root.val != right_root.val:
            return False
        left = self.mirror(left_root.left,right_root.right)
        right = self.mirror(left_root.right,right_root.left)
        return left and right


    def isSymmetric(self, root) -> bool:
        return self.mirror(root,root)

t = TreeNode(1)
t.left = TreeNode(2)
t.right = TreeNode(2)

t.left.left = TreeNode(3)
t.left.right = TreeNode(4)

t.right.left = TreeNode(4)
t.right.right = TreeNode(3)

r = Solution()
print(r.isSymmetric(t))
        
```

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    
    def invertTree(self, root):
        if root is None:
            return None

        from collections import deque
        queue = deque([(root,root.left,root.right)])

        while queue:
            node = queue.popleft()
            if node[1] is None and node[2] is None:
                continue
        
            node[0].left = node[2]
            node[0].right = node[1]
            if node[1]:
                queue.append((node[1],node[1].left,node[1].right))
            if node[2]:
                queue.append((node[2],node[2].left,node[2].right))

        return root 
        

        

t = TreeNode(4)
t.left = TreeNode(2)
t.right = TreeNode(7)

t.left.left = TreeNode(1)
t.left.right = TreeNode(3)

t.right.left = TreeNode(6)
t.right.right = TreeNode(9)

r = Solution()
r = r.invertTree(t)
        
```

#### Leetcode 110. Balanced Binary Tree

Given a binary tree, determine if it is height-balanced.

 

Example 1:


Input: root = [3,9,20,null,null,15,7]
Output: true
Example 2:


Input: root = [1,2,2,3,3,null,null,4,4]
Output: false
Example 3:

Input: root = []
Output: true
 

Constraints:

The number of nodes in the tree is in the range [0, 5000].
-104 <= Node.val <= 104

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def helper(self,root):
        if root is None:
            return (0,True)
        Balanced = False
        left = self.helper(root.left)
        right = self.helper(root.right)
        if left[1] and right[1]:
            if abs(left[0]-right[0]) <= 1:
                Balanced = True

        max_height = max(left[0],right[0]) + 1
        return (max_height,Balanced)
    
    def isBalanced(self, root) -> bool:
        max_height_and_balanced = self.helper(root)
        return max_height_and_balanced[1]

        
        

        

t = TreeNode(4)
t.left = TreeNode(2)
t.left.left = TreeNode(1)


r = Solution()
print(r.isBalanced(t))
        
```

Given the root of a binary tree, return the length of the diameter of the tree.

The diameter of a binary tree is the length of the longest path between any two nodes in a tree. This path may or may not pass through the root.

The length of a path between two nodes is represented by the number of edges between them.

 

Example 1:


Input: root = [1,2,3,4,5]
Output: 3
Explanation: 3 is the length of the path [4,2,1,3] or [5,2,1,3].

Note: The merging process must start from the root nodes of both trees.

 

Example 1:


Input: root1 = [1,3,2,5], root2 = [2,1,3,null,4,null,7]

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def mergeTrees(self, root1, root2):
        if root1 is None:
            return root2
        if root2 is None:
            return root1
        root1.val = root1.val + root2.val
            
        if root1.left is None and root2 and root2.left:
            root1.left = root2.left
        if root1.right is None and root2 and root2.right:
            root1.right = root2.right

        if root1.left != root2.left and root2.left:
            self.mergeTrees(root1.left,root2.left)
        if root1.right != root2.right and root2.right:
            self.mergeTrees(root1.right,root2.right)    

        return root1





t = TreeNode(3)
t.left = TreeNode(4)
t.left.left = TreeNode(1)



x = TreeNode(4)
x.right = TreeNode(2)


r = Solution()
r = r.mergeTrees(t,x)
```

A binary search tree is a binary tree where for every node, any descendant of Node.left has a value strictly less than Node.val, and any descendant of Node.right has a value strictly greater than Node.val.

A preorder traversal of a binary tree displays the value of the node first, then traverses Node.left, then traverses Node.right.

Input: preorder = [8,5,1,7,10,12]

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:  
    def small_next(self):
        node = self.small_stack.pop()
        if node.right:
            temp = node.right
            while temp:
                self.small_stack.append(temp)
                temp = temp.left
        return node.val

    def large_prev(self):
        node = self.large_stack.pop()
        if node.left:
            temp = node.left
            while temp:
                self.large_stack.append(temp)
                temp = temp.right
        return node.val

        

    
    def findTarget(self, root, k: int) -> bool:
        self.small_stack = []
        self.large_stack = []
        temp1 = root
        while temp1:
            self.small_stack.append(temp1)
            temp1 = temp1.left

        temp2 = root
        while temp2:
            self.large_stack.append(temp2)
            temp2 = temp2.right

        small = self.small_stack[-1].val
        large = self.large_stack[-1].val

        while small < large:

            total = small + large

            if total == k:
                return True

            if total < k:
                small = self.small_next()
            else:
                large = self.large_prev()

        return False


            
        
        
  


        
root = TreeNode(1)
node2 = TreeNode(0)
node3 = TreeNode(4)
node4 = TreeNode(-2)
node5 = TreeNode(3)



# 2. Connect the nodes according to the structure
root.left = node2
root.right = node3

node2.left = node4

node3.left = node5


r = Solution()
    
print(r.findTarget(root,k=7))
```

```python
# 2nd apporach
class Solution:
    def getMinimumDifference(self, root):
        prev = [None]
        minimum = [float("inf")]

        def inorder(node):
            if node is None:
                return

            # Left
            inorder(node.left)

            # Process current node
            if prev[0] is not None:
                diff = node.val - prev[0]
                minimum[0] = min(minimum[0], diff)

            # Update previous
            prev[0] = node.val

            # Right
            inorder(node.right)

        inorder(root)

        return minimum[0]
```
