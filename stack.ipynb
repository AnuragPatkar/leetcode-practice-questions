```python
class stack:
    def __init__(self):
        self._items = []

    def push(self,num):
        self._items.append(num)

    def peek(self):
        if not self.is_empty():
            return self._items[-1]
        raise IndexError("The stack is empty")


    def pop(self):
        if not self.is_empty():
            return self._items.pop()
        raise IndexError("The stack is empty")

    def is_empty(self):
        return len(self._items) == 0

    def size(self):
        return len(self._items)


s = stack()

print(s.is_empty())
s.push(10)
s.push(20)
s.push(30)
print(s.peek())
print(s.pop())
print(s.peek())
```
