Input: head = [1], pos = -1
Output: no cycle
Explanation: There is no cycle in the linked list.

```python
class Solution:
    def detectCycle(self,head):
       
        slow = head
        fast = head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
          
            if slow == fast:
                new_pointer = head
                while new_pointer != slow:
                    slow = slow.next
                    new_pointer = new_pointer.next
                return new_pointer
            
        return None
```
