#### Singly Linked List

```python
class Node:
    def __init__(self,value):
        self.value = value
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def insert_at_beginning(self,value):
        new_node = Node(value)
        if self.is_empty():
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head = new_node
        return True

    def insert_at_end(self,value):
        new_node = Node(value)
        if self.is_empty():
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node
        return True

    def delete_the_head(self):
        if self.is_empty():
            return False
        elif self.head.next is None and self.tail.next is None:
            self.head = None
            self.tail = None
        else:
            self.head =self.head.next
        return True

    def delete_the_tail(self):
        if self.is_empty():
            return False
        elif self.head.next is None and self.tail.next is None:
            self.head = None
            self.tail = None
        else:
            temp = self.head
            while temp.next.next is not None:
                temp = temp.next
            temp.next = None
            self.tail = temp
        return True

    def delete_node(self,value):
        if self.is_empty():
            return False
        elif self.head.value == value:
            return self.delete_the_head()
        elif self.tail.value == value:
            return self.delete_the_tail()
        else:
            temp = self.head.next
            previous_node = self.head
            while temp is not None:
                if temp.value == value:
                    previous_node.next = temp.next
                    return True
                previous_node = temp 
                temp = temp.next

        return False

    def insert_after(self,target,new_value):
        if self.is_empty():
            return False
        elif self.tail.value == target:
            return self.insert_at_end(new_value)
        temp = self.head
        while temp and temp.value != target:
            temp = temp.next
        if temp is None:
            return False
        else:
            new_node = Node(new_value)
            new_node.next = temp.next
            temp.next = new_node
            return True

    def insert_before(self,target,new_value):
        if self.is_empty():
            return False
        elif self.head.value == target:
            return self.insert_at_beginning(new_value)
        else:
            temp = self.head.next
            previous_node = self.head
            while temp and temp.value != target:
                previous_node = temp
                temp = temp.next
            if temp is None:
                return False
            else:
                new_node = Node(new_value)
                new_node.next = previous_node.next
                previous_node.next = new_node
                return True


    
    def display(self):
        temp = self.head
        while temp is not None:
            print(temp.value)
            temp = temp.next

    def search(self,value):
        temp = self.head
        while temp is not None:
            if temp.value == value:
                return True
            temp = temp.next
        return False
    
    def is_empty(self):
        return self.head is None


l = LinkedList()
print(l.is_empty())
l.insert_at_beginning(10)
l.insert_at_beginning(20)
l.insert_at_beginning(30)
print(l.head.value)
print(l.tail.value)
l.insert_at_end(40)
print(l.tail.value)
l.display()
```
